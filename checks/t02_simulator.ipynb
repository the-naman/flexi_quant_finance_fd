```python
import collections, os, shutil, subprocess, sys

root = os.path.dirname(os.getcwd())
out = "/tmp/sim_test"
shutil.rmtree(out, ignore_errors=True)
p = subprocess.run([sys.executable, "-m", "simulator.run", "--company", "fqf", "--env", "dev",
                    "--out", out, "--upto", "2026-10-07"], cwd=root, capture_output=True, text=True)
print(p.stdout, p.stderr[-1500:])

got = collections.Counter()
for folder, _, files in os.walk(out + "/landing"):
    if "/_sim" in folder:
        continue
    for f in files:
        got[folder.split("/landing/")[1].split("/")[0] + " ." + f.rsplit(".", 1)[1]] += 1

want = {"bank .pdf": 14, "branch_scans .png": 12, "bureau_api .json": 17, "excel .csv": 17,
        "gateway .json": 17, "legacy .txt": 51, "mysql .csv": 34, "salesforce .csv": 51}
print(dict(sorted(got.items())))
print("TOTAL", sum(got.values()), "PASS" if dict(got) == want else "FAIL")
```
