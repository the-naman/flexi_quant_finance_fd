# One-time setup · fqf_prod
Creates the catalog, schemas and volumes for fqf_prod. Run once by hand. Not part of any job.

```sql
CREATE CATALOG IF NOT EXISTS fqf_prod COMMENT 'Flexi Quant Finance, prod environment';
CREATE SCHEMA IF NOT EXISTS fqf_prod.bronze COMMENT 'raw data as landed';
CREATE SCHEMA IF NOT EXISTS fqf_prod.silver COMMENT 'clean, validated, SCD2 history';
CREATE SCHEMA IF NOT EXISTS fqf_prod.gold   COMMENT 'business-ready facts and dimensions';
CREATE SCHEMA IF NOT EXISTS fqf_prod.ops    COMMENT 'audit, watermarks, quality results';
CREATE VOLUME IF NOT EXISTS fqf_prod.ops.checkpoints;
CREATE VOLUME IF NOT EXISTS fqf_prod.ops.refresh_out;
```
