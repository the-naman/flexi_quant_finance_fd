# Grant access (one-time, by hand)
- Gives each role group (and pipeline service principal) its access to one environment's catalog, per the access matrix in `src/lh/access.py` (also checked by `02_verify_setup`).
- **How:** Environment → Mode = preview → Run all → check → Mode = apply → Run all. Safe to re-run. Not part of any job.
- Masking (column masks) is added on tables in Phase 5. Ownership transfer is a separate later step.

## 2. Repo constant and widgets

```python
import os, sys
sys.path.insert(0, os.path.join(os.path.dirname(os.path.dirname(os.getcwd())), "src"))
import yaml
from lh import access, common
REPO_COMPANY = yaml.safe_load(open(f"{common.repo_root()}/config/company.yml"))["code"]   # bhg or fqf

dbutils.widgets.dropdown("env",  "dev",     ["dev", "test", "prod"], "Environment")
dbutils.widgets.dropdown("mode", "preview", ["preview", "apply"],    "Mode")
```

## 3. Access matrix

```python
env     = dbutils.widgets.get("env")
mode    = dbutils.widgets.get("mode")
catalog = f"{REPO_COMPANY}_{env}"
SP      = f"sp_{REPO_COMPANY}_{env}"

# The access matrix lives in src/lh/access.py (same file in both repos), so this notebook and 02_verify_setup
# always agree. Raw files (READ/WRITE VOLUME) only for roles that may see real values; lh_ci = automation
# (sp_lh_notices) reads Gold and run logs; FQF test/prod: the BHG job account reads Gold for the group view.
FILES  = access.FILES
MATRIX = {e: access.matrix(REPO_COMPANY, e) for e in ("dev", "test", "prod")}
print(f"catalog={catalog} | mode={mode} | {len(MATRIX[env])} principals")
```

## 4. Resolve names (service principals are granted by their application ID)

```python
from databricks.sdk import WorkspaceClient
w = WorkspaceClient()

def resolve(name):
    if name.startswith("sp_"):
        sps = list(w.service_principals.list(filter=f"displayName eq '{name}'"))
        if not sps:
            raise Exception(f"Service principal '{name}' not found")
        return sps[0].application_id
    if not list(w.groups.list(filter=f"displayName eq '{name}'")):
        raise Exception(f"Group '{name}' not found")
    return name

IDS = {name: resolve(name) for name, _, _ in MATRIX[env]}
NAME_OF = {v: k for k, v in IDS.items()}
print("Resolved:", ", ".join(IDS))
```

## 5. Plan

```python
granted = {(r.Principal, r.ActionType.replace("_", " ").upper())
           for r in spark.sql(f"SHOW GRANTS ON CATALOG {catalog}").collect() if r.ObjectType.upper() == "CATALOG"}

stmts = []
for name, scope, privs in MATRIX[env]:
    p = IDS[name]
    if scope == "catalog":
        stmts.append((name, f"GRANT {', '.join(privs)} ON CATALOG {catalog} TO `{p}`"))
        gone = [x for x in FILES if x not in privs and (p, x) in granted]     # take away file access no longer in the matrix
        if gone:
            stmts.append((name, f"REVOKE {', '.join(gone)} ON CATALOG {catalog} FROM `{p}`"))
    else:  # one schema (gold or ops): catalog USE/BROWSE once per principal
        cat_grant = (name, f"GRANT USE CATALOG, BROWSE ON CATALOG {catalog} TO `{p}`")
        if cat_grant not in stmts:
            stmts.append(cat_grant)
        stmts.append((name, f"GRANT {', '.join(privs)} ON SCHEMA {catalog}.{scope} TO `{p}`"))

for name, s in stmts:
    print(f"[{name}] {s.replace(IDS[name], name) if name.startswith('sp_') else s};")   # names only, never an id
print(f"\nPlan: {len(stmts)} statement(s), {sum(s.startswith('REVOKE') for _, s in stmts)} of them REVOKE")
```

## 6. Apply

```python
if mode == "apply":
    for _, s in stmts:
        spark.sql(s)
    print(f"Applied {len(stmts)} statement(s) to {catalog}")
else:
    print("Preview only. Set Mode = apply and Run all.")
```

## 7. Verify (direct grants only), with Missing and Extra

```python
norm = lambda a: a.replace("_", " ").upper()
# expected = access matrix + the controlled-refresh grants of 03_refresh_setup (both in src/lh/access.py)
for name, _, _ in access.refresh_grants(REPO_COMPANY, env):
    if name not in IDS:
        IDS[name] = resolve(name)
NAME_OF = {v: k for k, v in IDS.items()}

def direct_grants(obj_type, obj):
    rows = spark.sql(f"SHOW GRANTS ON {obj_type} {obj}").collect()
    who = lambda p: "a person" if "@" in p else NAME_OF.get(p, p)          # never print an address
    return {(who(r.Principal), norm(r.ActionType)) for r in rows if r.ObjectType.upper() == obj_type}

want = {k: {(n, norm(p)) for n, p in v} for k, v in access.expected(REPO_COMPANY, env).items()}
want.setdefault(("SCHEMA", f"{catalog}.gold"), set())
for (kind, obj), expected in sorted(want.items()):
    actual = direct_grants(kind, obj)
    label = f"{kind} {obj}"
    print(f"{label:40} missing:", sorted(expected - actual) or "none")
    print(f"{label:40} extra  :", sorted(actual - expected) or "none")
```
