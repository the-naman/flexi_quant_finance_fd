# One-time setup · fqf_test
Creates the catalog, schemas and volumes for fqf_test. Run once by hand. Not part of any job.

```sql
CREATE CATALOG IF NOT EXISTS fqf_test COMMENT 'Flexi Quant Finance, test environment';
CREATE SCHEMA IF NOT EXISTS fqf_test.bronze COMMENT 'raw data as landed';
CREATE SCHEMA IF NOT EXISTS fqf_test.silver COMMENT 'clean, validated, SCD2 history';
CREATE SCHEMA IF NOT EXISTS fqf_test.gold   COMMENT 'business-ready facts and dimensions';
CREATE SCHEMA IF NOT EXISTS fqf_test.ops    COMMENT 'audit, watermarks, quality results';
CREATE VOLUME IF NOT EXISTS fqf_test.bronze.landing;
CREATE VOLUME IF NOT EXISTS fqf_test.ops.checkpoints;
CREATE VOLUME IF NOT EXISTS fqf_test.ops.refresh_in;
CREATE VOLUME IF NOT EXISTS fqf_test.ops.refresh_out;
```
