# One-time setup · fqf_dev
Creates the catalog, schemas and volumes for fqf_dev. Run once by hand. Not part of any job.

```sql
CREATE CATALOG IF NOT EXISTS fqf_dev COMMENT 'Flexi Quant Finance, dev environment';
CREATE SCHEMA IF NOT EXISTS fqf_dev.bronze COMMENT 'raw data as landed';
CREATE SCHEMA IF NOT EXISTS fqf_dev.silver COMMENT 'clean, validated, SCD2 history';
CREATE SCHEMA IF NOT EXISTS fqf_dev.gold   COMMENT 'business-ready facts and dimensions';
CREATE SCHEMA IF NOT EXISTS fqf_dev.ops    COMMENT 'audit, watermarks, quality results';
CREATE VOLUME IF NOT EXISTS fqf_dev.ops.checkpoints;
CREATE VOLUME IF NOT EXISTS fqf_dev.ops.refresh_in;
```
