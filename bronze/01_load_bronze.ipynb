```python
import os, sys
sys.path.insert(0, os.path.join(os.path.dirname(os.getcwd()), "src"))
from lh import common, bronze

dbutils.widgets.dropdown("env", "dev", ["dev", "test", "prod"])
env = dbutils.widgets.get("env")
cfg = common.load_config(env)
key = common.match_key(env)
tables = bronze.load_tables()
# table: "all" = every table (the daily job always runs this); one name = only that table, for developing or
# fixing a single table. Each table keeps its own checkpoint, files, audit row (bronze.<table>) and checks.
dbutils.widgets.dropdown("table", "all", ["all"] + sorted(tables))
pick = dbutils.widgets.get("table")
if pick != "all":
    tables = {pick: tables[pick]}
print(cfg["code"], env, len(tables), "tables" if pick == "all" else f"table (only {pick})")
```

```python
landing = bronze.list_landing(spark, cfg)
import time
start = time.time()
done = bronze.load_all(spark, cfg, tables, key, landing)     # tables side by side (config/performance.yml)
total = loaded = rejected = 0
for table, (rows, ok, bad) in done.items():
    total, loaded, rejected = total + rows, loaded + ok, rejected + bad
    print(f"{table:20} +{rows:<6} files {ok:<4} rejected {bad}")
print(f"Bronze took {time.time() - start:.0f}s")
if pick == "all":                                    # one table: its own bronze.<table> row is the audit
    common.log_run(spark, cfg, "bronze", "success", total, f"{loaded} files loaded, {rejected} rejected")
print("TOTAL rows", total, "| files loaded", loaded, "| rejected", rejected)
```
